```python
# ======================================================================
# AI_TEXT_GEN - STANDALONE SIZE REDUCTION + ZIP CREATOR
#
# Run this cell by itself.
#
# It DOES NOT require PROJECT_DIR to have been defined previously.
# It DOES NOT retrain your model.
#
# Target:
# C:\Users\sagni\Downloads\AI_Text_gen
#
# Goal:
# Create submission ZIP below 50 MB
# ======================================================================


# ======================================================================
# STEP 1: IMPORT LIBRARIES
# ======================================================================

import os
import zipfile
import shutil
import json
from pathlib import Path


# ======================================================================
# STEP 2: DEFINE ALL PATHS
#
# THIS FIXES YOUR NameError
# ======================================================================

PROJECT_DIR = r"C:\Users\sagni\Downloads\AI_Text_gen"

CHECKPOINT_DIR = os.path.join(
    PROJECT_DIR,
    "checkpoints"
)

RESULTS_DIR = os.path.join(
    PROJECT_DIR,
    "results"
)

SAMPLES_DIR = os.path.join(
    PROJECT_DIR,
    "generated_samples"
)

ZIP_PATH = os.path.join(
    os.path.dirname(PROJECT_DIR),
    "AI_Text_gen_submission.zip"
)


# ======================================================================
# STEP 3: CHECK PROJECT FOLDER
# ======================================================================

if not os.path.exists(PROJECT_DIR):

    raise FileNotFoundError(
        f"\nProject folder does not exist:\n"
        f"{PROJECT_DIR}"
    )


print(
    "=" * 75
)

print(
    "AI TEXT GENERATION - SUBMISSION SIZE OPTIMIZER"
)

print(
    "=" * 75
)

print(
    "\nProject folder:"
)

print(
    PROJECT_DIR
)


# ======================================================================
# STEP 4: CREATE FOLDERS IF NEEDED
# ======================================================================

os.makedirs(
    CHECKPOINT_DIR,
    exist_ok=True
)

os.makedirs(
    RESULTS_DIR,
    exist_ok=True
)

os.makedirs(
    SAMPLES_DIR,
    exist_ok=True
)


# ======================================================================
# STEP 5: SIZE HELPER
# ======================================================================

def bytes_to_mb(size):

    return (
        size
        / (1024 * 1024)
    )


def get_folder_size(folder):

    total_size = 0

    for root, dirs, files in os.walk(folder):

        for filename in files:

            filepath = os.path.join(
                root,
                filename
            )

            try:

                total_size += os.path.getsize(
                    filepath
                )

            except OSError:

                pass

    return total_size


# ======================================================================
# STEP 6: SHOW ORIGINAL FOLDER SIZE
# ======================================================================

original_size = get_folder_size(
    PROJECT_DIR
)


print(
    f"\nOriginal folder size: "
    f"{bytes_to_mb(original_size):.2f} MB"
)


# ======================================================================
# STEP 7: DISPLAY LARGE FILES
# ======================================================================

def get_all_files(folder):

    information = []

    for root, dirs, files in os.walk(folder):

        for filename in files:

            filepath = os.path.join(
                root,
                filename
            )

            try:

                size = os.path.getsize(
                    filepath
                )

            except OSError:

                continue

            relative_path = os.path.relpath(
                filepath,
                folder
            )

            information.append(
                (
                    filepath,
                    relative_path,
                    size
                )
            )

    information.sort(
        key=lambda item: item[2],
        reverse=True
    )

    return information


print(
    "\n"
    + "=" * 75
)

print(
    "LARGEST FILES BEFORE CLEANUP"
)

print(
    "=" * 75
)


files_before = get_all_files(
    PROJECT_DIR
)


for filepath, relative_path, size in files_before[:30]:

    print(
        f"{bytes_to_mb(size):8.2f} MB   "
        f"{relative_path}"
    )


# ======================================================================
# STEP 8: FILES WE WANT TO KEEP
# ======================================================================

# Required checkpoints according to assignment:
#
# Epoch 10
# Epoch 20
# Epoch 30
# Final model

required_checkpoint_numbers = {
    "10",
    "20",
    "30"
}


# ======================================================================
# STEP 9: DELETE UNNECESSARY CHECKPOINTS
#
# Removes:
# epoch 5
# epoch 15
# epoch 25
#
# Keeps:
# epoch 10
# epoch 20
# epoch 30
# ======================================================================

print(
    "\n"
    + "=" * 75
)

print(
    "CHECKPOINT CLEANUP"
)

print(
    "=" * 75
)


if os.path.exists(CHECKPOINT_DIR):

    checkpoint_files = os.listdir(
        CHECKPOINT_DIR
    )

    for filename in checkpoint_files:

        filepath = os.path.join(
            CHECKPOINT_DIR,
            filename
        )

        filename_lower = filename.lower()

        # --------------------------------------------------------------
        # Identify unnecessary epochs
        # --------------------------------------------------------------

        unnecessary = False

        if "epoch_05" in filename_lower:

            unnecessary = True

        elif "epoch_5." in filename_lower:

            unnecessary = True

        elif "epoch_15" in filename_lower:

            unnecessary = True

        elif "epoch_25" in filename_lower:

            unnecessary = True

        # --------------------------------------------------------------
        # Delete unnecessary checkpoint
        # --------------------------------------------------------------

        if unnecessary:

            try:

                size = os.path.getsize(
                    filepath
                )

                os.remove(
                    filepath
                )

                print(
                    f"Removed "
                    f"{bytes_to_mb(size):.2f} MB: "
                    f"{filename}"
                )

            except Exception as error:

                print(
                    "Could not remove:",
                    filepath
                )

                print(
                    error
                )


# ======================================================================
# STEP 10: REMOVE GRU MODEL WEIGHTS
#
# Your assignment comparison only needs the GRU results/metrics.
#
# Keep:
# lstm_vs_gru_comparison.csv
# lstm_gru_comparison.png
#
# Remove:
# gru_final.weights.h5
# ======================================================================

GRU_WEIGHT_FILES = [

    os.path.join(
        PROJECT_DIR,
        "gru_final.weights.h5"
    ),

    os.path.join(
        PROJECT_DIR,
        "gru_model.h5"
    ),

    os.path.join(
        PROJECT_DIR,
        "gru_final.h5"
    )
]


for filepath in GRU_WEIGHT_FILES:

    if os.path.exists(filepath):

        size = os.path.getsize(
            filepath
        )

        try:

            os.remove(
                filepath
            )

            print(
                f"\nRemoved duplicate GRU model: "
                f"{bytes_to_mb(size):.2f} MB"
            )

            print(
                filepath
            )

        except Exception as error:

            print(
                "\nCould not remove:"
            )

            print(
                filepath
            )

            print(
                error
            )


# ======================================================================
# STEP 11: REMOVE FULL LSTM MODEL IF FINAL WEIGHTS EXIST
#
# If lstm_final.weights.h5 exists, then
# character_lstm_model.h5 is another large copy.
#
# We keep the final weights and remove the full model.
# ======================================================================

FINAL_LSTM_WEIGHTS = os.path.join(
    PROJECT_DIR,
    "lstm_final.weights.h5"
)

FULL_LSTM_MODEL = os.path.join(
    PROJECT_DIR,
    "character_lstm_model.h5"
)


if (
    os.path.exists(FINAL_LSTM_WEIGHTS)
    and
    os.path.exists(FULL_LSTM_MODEL)
):

    full_model_size = os.path.getsize(
        FULL_LSTM_MODEL
    )

    try:

        os.remove(
            FULL_LSTM_MODEL
        )

        print(
            f"\nRemoved duplicate full LSTM model: "
            f"{bytes_to_mb(full_model_size):.2f} MB"
        )

        print(
            FULL_LSTM_MODEL
        )

    except Exception as error:

        print(
            "\nCould not remove full LSTM model:"
        )

        print(
            error
        )


# ======================================================================
# STEP 12: REMOVE TEMPORARY/JUNK FILES
# ======================================================================

unwanted_extensions = {

    ".tmp",
    ".temp",
    ".bak",
    ".log~"
}


unwanted_names = {

    "thumbs.db",
    ".ds_store"
}


removed_junk_size = 0


for root, dirs, files in os.walk(
    PROJECT_DIR
):

    for filename in files:

        filepath = os.path.join(
            root,
            filename
        )

        extension = os.path.splitext(
            filename
        )[1].lower()

        should_remove = (
            extension in unwanted_extensions
            or
            filename.lower() in unwanted_names
        )

        if should_remove:

            try:

                file_size = os.path.getsize(
                    filepath
                )

                os.remove(
                    filepath
                )

                removed_junk_size += (
                    file_size
                )

            except OSError:

                pass


if removed_junk_size > 0:

    print(
        f"\nRemoved junk files: "
        f"{bytes_to_mb(removed_junk_size):.2f} MB"
    )


# ======================================================================
# STEP 13: REMOVE __PYCACHE__
# ======================================================================

for root, dirs, files in os.walk(
    PROJECT_DIR,
    topdown=False
):

    for directory in dirs:

        if directory == "__pycache__":

            cache_path = os.path.join(
                root,
                directory
            )

            try:

                shutil.rmtree(
                    cache_path
                )

                print(
                    "\nRemoved cache:"
                )

                print(
                    cache_path
                )

            except OSError:

                pass


# ======================================================================
# STEP 14: CHECK FOR ACCIDENTAL ZIP FILES INSIDE PROJECT
#
# Very important:
#
# Do not keep old ZIP files INSIDE AI_Text_gen because then the new
# submission ZIP will contain another ZIP and become much larger.
# ======================================================================

for root, dirs, files in os.walk(
    PROJECT_DIR
):

    for filename in files:

        if filename.lower().endswith(
            ".zip"
        ):

            filepath = os.path.join(
                root,
                filename
            )

            try:

                size = os.path.getsize(
                    filepath
                )

                os.remove(
                    filepath
                )

                print(
                    f"\nRemoved nested ZIP "
                    f"({bytes_to_mb(size):.2f} MB):"
                )

                print(
                    filepath
                )

            except OSError:

                pass


# ======================================================================
# STEP 15: SHOW SIZE AFTER CLEANUP
# ======================================================================

cleaned_size = get_folder_size(
    PROJECT_DIR
)


print(
    "\n"
    + "=" * 75
)

print(
    "SIZE AFTER CLEANUP"
)

print(
    "=" * 75
)


print(
    f"\nBefore: "
    f"{bytes_to_mb(original_size):.2f} MB"
)


print(
    f"After : "
    f"{bytes_to_mb(cleaned_size):.2f} MB"
)


print(
    f"Saved : "
    f"{bytes_to_mb(original_size - cleaned_size):.2f} MB"
)


# ======================================================================
# STEP 16: SHOW REMAINING LARGE FILES
# ======================================================================

print(
    "\n"
    + "=" * 75
)

print(
    "LARGEST REMAINING FILES"
)

print(
    "=" * 75
)


files_after = get_all_files(
    PROJECT_DIR
)


for filepath, relative_path, size in files_after[:30]:

    print(
        f"{bytes_to_mb(size):8.2f} MB   "
        f"{relative_path}"
    )


# ======================================================================
# STEP 17: CHECK IMPORTANT FILES
# ======================================================================

print(
    "\n"
    + "=" * 75
)

print(
    "IMPORTANT SUBMISSION FILES"
)

print(
    "=" * 75
)


important_files = [

    "shakespeare.txt",

    "char2idx.json",

    "idx2char.json",

    "dataset_info.json",

    "training_history.csv",

    "training_history.json",

    "lstm_model_summary.txt",

    "gru_model_summary.txt",

    "final_metrics.json",

    "analysis_report.txt",

    "accuracy_graph.png",

    "training_validation_loss.png",

    "perplexity_graph.png",

    "lstm_gru_comparison.png",

    "character_bigram_comparison.png",

    "hidden_state_tsne.png",

    "temperature_effect_comparison.png"
]


for filename in important_files:

    filepath = os.path.join(
        PROJECT_DIR,
        filename
    )

    if os.path.exists(filepath):

        print(
            "[OK]     ",
            filename
        )

    else:

        print(
            "[MISSING]",
            filename
        )


# ======================================================================
# STEP 18: CHECK REQUIRED CHECKPOINTS
# ======================================================================

print(
    "\nRequired checkpoint status:"
)


for epoch in [
    10,
    20,
    30
]:

    possible_names = [

        f"lstm_epoch_{epoch:02d}.weights.h5",

        f"lstm_epoch_{epoch}.weights.h5",

        f"lstm_epoch_{epoch:02d}_compressed.npz",

        f"lstm_epoch_{epoch}_compressed.npz"
    ]


    found = False


    for filename in possible_names:

        filepath = os.path.join(
            CHECKPOINT_DIR,
            filename
        )

        if os.path.exists(filepath):

            size = os.path.getsize(
                filepath
            )

            print(
                f"[OK] Epoch {epoch}: "
                f"{filename} "
                f"({bytes_to_mb(size):.2f} MB)"
            )

            found = True

            break


    if not found:

        print(
            f"[MISSING] Epoch {epoch}"
        )


# ======================================================================
# STEP 19: FINAL MODEL CHECK
# ======================================================================

possible_final_models = [

    os.path.join(
        PROJECT_DIR,
        "lstm_final.weights.h5"
    ),

    os.path.join(
        PROJECT_DIR,
        "lstm_final_compressed.npz"
    ),

    os.path.join(
        PROJECT_DIR,
        "character_lstm_model.h5"
    )
]


final_found = False


for filepath in possible_final_models:

    if os.path.exists(filepath):

        size = os.path.getsize(
            filepath
        )

        print(
            f"\n[OK] Final LSTM model: "
            f"{os.path.basename(filepath)} "
            f"({bytes_to_mb(size):.2f} MB)"
        )

        final_found = True

        break


if not final_found:

    print(
        "\n[WARNING] No final LSTM weight/model file found."
    )


# ======================================================================
# STEP 20: CREATE NEW ZIP
# ======================================================================

if os.path.exists(
    ZIP_PATH
):

    try:

        os.remove(
            ZIP_PATH
        )

    except OSError:

        pass


print(
    "\n"
    + "=" * 75
)

print(
    "CREATING FINAL SUBMISSION ZIP"
)

print(
    "=" * 75
)


print(
    "\nOutput:"
)

print(
    ZIP_PATH
)


with zipfile.ZipFile(
    ZIP_PATH,
    mode="w",
    compression=zipfile.ZIP_DEFLATED,
    compresslevel=9
) as zip_file:

    for root, dirs, files in os.walk(
        PROJECT_DIR
    ):

        for filename in files:

            filepath = os.path.join(
                root,
                filename
            )

            archive_name = os.path.relpath(
                filepath,
                os.path.dirname(
                    PROJECT_DIR
                )
            )

            zip_file.write(
                filepath,
                archive_name
            )


# ======================================================================
# STEP 21: FINAL ZIP SIZE
# ======================================================================

zip_size = os.path.getsize(
    ZIP_PATH
)


zip_size_mb = bytes_to_mb(
    zip_size
)


print(
    "\n"
    + "=" * 75
)

print(
    "FINAL RESULT"
)

print(
    "=" * 75
)


print(
    f"\nProject folder size : "
    f"{bytes_to_mb(cleaned_size):.2f} MB"
)


print(
    f"Compressed ZIP size : "
    f"{zip_size_mb:.2f} MB"
)


print(
    "\nZIP location:"
)

print(
    ZIP_PATH
)


# ======================================================================
# STEP 22: RESULT MESSAGE
# ======================================================================

if zip_size_mb < 50:

    print(
        "\n"
        + "=" * 75
    )

    print(
        "SUCCESS"
    )

    print(
        "=" * 75
    )

    print(
        f"\nYour ZIP is {zip_size_mb:.2f} MB."
    )

    print(
        "It is below the 50 MB submission limit."
    )


else:

    print(
        "\n"
        + "=" * 75
    )

    print(
        "ZIP IS STILL ABOVE 50 MB"
    )

    print(
        "=" * 75
    )

    print(
        f"\nCurrent ZIP: "
        f"{zip_size_mb:.2f} MB"
    )

    print(
        "\nLargest remaining files:"
    )


    for filepath, relative_path, size in files_after[:15]:

        print(
            f"{bytes_to_mb(size):8.2f} MB   "
            f"{relative_path}"
        )


    print(
        "\nIf it is only slightly above 50 MB, "
        "the remaining model checkpoints are probably responsible."
    )


# ======================================================================
# END
# ======================================================================
```
