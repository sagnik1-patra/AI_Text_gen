```python
import os
import json
import math
import random
import zipfile
import urllib.request
import warnings
from collections import Counter

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.manifold import TSNE

import tensorflow as tf

from tensorflow.keras import (
    layers,
    models,
    callbacks,
    optimizers
)


# ======================================================================
# STEP 2: REPRODUCIBILITY
# ======================================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)


# ======================================================================
# STEP 3: PROJECT DIRECTORY
# ======================================================================

PROJECT_DIR = r"C:\Users\sagni\Downloads\AI_Text_gen"

CHECKPOINT_DIR = os.path.join(
    PROJECT_DIR,
    "checkpoints"
)

SAMPLES_DIR = os.path.join(
    PROJECT_DIR,
    "generated_samples"
)

RESULTS_DIR = os.path.join(
    PROJECT_DIR,
    "results"
)


os.makedirs(
    PROJECT_DIR,
    exist_ok=True
)

os.makedirs(
    CHECKPOINT_DIR,
    exist_ok=True
)

os.makedirs(
    SAMPLES_DIR,
    exist_ok=True
)

os.makedirs(
    RESULTS_DIR,
    exist_ok=True
)


# ======================================================================
# STEP 4: CONFIGURATION
# ======================================================================

SEQUENCE_LENGTH = 100

STRIDE = 3

EMBEDDING_DIM = 128

HIDDEN_SIZE = 256

DROPOUT_RATE = 0.30

BATCH_SIZE = 128

LEARNING_RATE = 0.001

MAX_GRADIENT_NORM = 5.0

GENERATE_LENGTH = 500


# ======================================================================
# IMPORTANT
#
# For testing:
# EPOCHS = 3
#
# For final assignment:
# EPOCHS = 30
# ======================================================================

EPOCHS = 30


print(
    "=" * 75
)

print(
    "CHARACTER-LEVEL RNN TEXT GENERATION"
)

print(
    "=" * 75
)

print(
    "\nProject Directory:"
)

print(
    PROJECT_DIR
)

print(
    "\nTensorFlow Version:",
    tf.__version__
)

print(
    "\nAvailable GPUs:"
)

print(
    tf.config.list_physical_devices("GPU")
)


# ======================================================================
# STEP 5: DOWNLOAD SHAKESPEARE DATASET
# ======================================================================

TEXT_FILE = os.path.join(
    PROJECT_DIR,
    "shakespeare.txt"
)


DATASET_URL = (
    "https://storage.googleapis.com/"
    "download.tensorflow.org/data/shakespeare.txt"
)


if not os.path.exists(TEXT_FILE):

    print(
        "\nDownloading Shakespeare dataset..."
    )

    urllib.request.urlretrieve(
        DATASET_URL,
        TEXT_FILE
    )

    print(
        "Dataset downloaded successfully."
    )

else:

    print(
        "\nDataset already exists."
    )


# ======================================================================
# STEP 6: LOAD DATASET
# ======================================================================

with open(
    TEXT_FILE,
    "r",
    encoding="utf-8"
) as file:

    text = file.read()


print(
    "\nDataset loaded successfully."
)

print(
    "Total characters:",
    len(text)
)


# ======================================================================
# STEP 7: BASIC TEXT ANALYSIS
# ======================================================================

character_counts = Counter(
    text
)

characters = sorted(
    character_counts.keys()
)

VOCAB_SIZE = len(
    characters
)


print(
    "Unique characters:",
    VOCAB_SIZE
)


# ======================================================================
# STEP 8: CHARACTER FREQUENCY TABLE
# ======================================================================

frequency_df = pd.DataFrame(
    {
        "Character": [
            repr(character)
            for character
            in characters
        ],

        "Frequency": [
            character_counts[
                character
            ]
            for character
            in characters
        ]
    }
)


frequency_df = frequency_df.sort_values(
    "Frequency",
    ascending=False
)


frequency_df.to_csv(
    os.path.join(
        RESULTS_DIR,
        "character_frequency.csv"
    ),
    index=False,
    encoding="utf-8"
)


print(
    "\nMost common characters:"
)

print(
    frequency_df.head(20)
)


# ======================================================================
# STEP 9: CHARACTER FREQUENCY GRAPH
# ======================================================================

top_frequency = frequency_df.head(
    40
)


plt.figure(
    figsize=(14, 7)
)

plt.bar(
    top_frequency["Character"],
    top_frequency["Frequency"]
)

plt.title(
    "Top 40 Character Frequencies in Shakespeare Corpus"
)

plt.xlabel(
    "Character"
)

plt.ylabel(
    "Frequency"
)

plt.xticks(
    rotation=90
)

plt.tight_layout()


plt.savefig(
    os.path.join(
        PROJECT_DIR,
        "character_frequency.png"
    ),
    dpi=150,
    bbox_inches="tight"
)

plt.show()

plt.close()


# ======================================================================
# STEP 10: CREATE CHARACTER MAPPINGS
# ======================================================================

char2idx = {
    character: index
    for index, character
    in enumerate(characters)
}


idx2char = {
    index: character
    for character, index
    in char2idx.items()
}


print(
    "\nVocabulary Size:",
    VOCAB_SIZE
)


# ======================================================================
# STEP 11: SAVE CHARACTER MAPPINGS
# ======================================================================

CHAR2IDX_PATH = os.path.join(
    PROJECT_DIR,
    "char2idx.json"
)


IDX2CHAR_PATH = os.path.join(
    PROJECT_DIR,
    "idx2char.json"
)


with open(
    CHAR2IDX_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        char2idx,
        file,
        indent=4,
        ensure_ascii=False
    )


with open(
    IDX2CHAR_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        {
            str(key): value
            for key, value
            in idx2char.items()
        },
        file,
        indent=4,
        ensure_ascii=False
    )


print(
    "\nVocabulary mappings saved."
)


# ======================================================================
# STEP 12: SAVE DATASET INFORMATION
# ======================================================================

dataset_info = {

    "dataset":
        "Shakespeare Complete Works",

    "total_characters":
        len(text),

    "unique_characters":
        VOCAB_SIZE,

    "sequence_length":
        SEQUENCE_LENGTH,

    "stride":
        STRIDE,

    "training_split":
        0.90,

    "validation_split":
        0.10,

    "embedding_dimension":
        EMBEDDING_DIM,

    "hidden_size":
        HIDDEN_SIZE,

    "batch_size":
        BATCH_SIZE,

    "learning_rate":
        LEARNING_RATE,

    "gradient_clip_norm":
        MAX_GRADIENT_NORM
}


with open(
    os.path.join(
        PROJECT_DIR,
        "dataset_info.json"
    ),
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        dataset_info,
        file,
        indent=4,
        ensure_ascii=False
    )


# ======================================================================
# STEP 13: ENCODE COMPLETE TEXT
# ======================================================================

encoded_text = np.array(
    [
        char2idx[
            character
        ]
        for character
        in text
    ],
    dtype=np.int32
)


print(
    "\nEncoded text shape:",
    encoded_text.shape
)


# ======================================================================
# STEP 14: CREATE TRAINING SEQUENCES
#
# Input:
# 100 characters
#
# Target:
# Next character
#
# Stride:
# 3
# ======================================================================

def create_sequences(
    encoded_data,
    sequence_length=100,
    stride=3
):

    X = []

    y = []


    for start in range(
        0,
        len(encoded_data) - sequence_length,
        stride
    ):

        end = (
            start
            + sequence_length
        )


        X.append(
            encoded_data[
                start:end
            ]
        )


        y.append(
            encoded_data[
                end
            ]
        )


    X = np.asarray(
        X,
        dtype=np.int32
    )


    y = np.asarray(
        y,
        dtype=np.int32
    )


    return X, y


print(
    "\nCreating training sequences..."
)


X, y = create_sequences(
    encoded_text,
    SEQUENCE_LENGTH,
    STRIDE
)


print(
    "Total sequences:",
    len(X)
)

print(
    "X shape:",
    X.shape
)

print(
    "y shape:",
    y.shape
)


# ======================================================================
# STEP 15: TRAIN / VALIDATION SPLIT
# ======================================================================

split_index = int(
    0.90
    * len(X)
)


X_train = X[
    :split_index
]


y_train = y[
    :split_index
]


X_val = X[
    split_index:
]


y_val = y[
    split_index:
]


print(
    "\nTraining samples:",
    len(X_train)
)

print(
    "Validation samples:",
    len(X_val)
)


# ======================================================================
# STEP 16: TF.DATA DATASETS
# ======================================================================

AUTOTUNE = tf.data.AUTOTUNE


train_dataset = (
    tf.data.Dataset
    .from_tensor_slices(
        (
            X_train,
            y_train
        )
    )
    .shuffle(
        min(
            10000,
            len(X_train)
        ),
        seed=SEED
    )
    .batch(
        BATCH_SIZE
    )
    .prefetch(
        AUTOTUNE
    )
)


validation_dataset = (
    tf.data.Dataset
    .from_tensor_slices(
        (
            X_val,
            y_val
        )
    )
    .batch(
        BATCH_SIZE
    )
    .prefetch(
        AUTOTUNE
    )
)


# ======================================================================
# STEP 17: BUILD LSTM MODEL
#
# Required:
#
# Embedding(vocab_size, 128)
# LSTM(256, return_sequences=True)
# Dropout(0.3)
# LSTM(256)
# Dense(vocab_size, softmax)
# ======================================================================

def build_lstm_model(
    vocab_size,
    sequence_length=100
):

    model = models.Sequential(
        [

            layers.Input(
                shape=(
                    sequence_length,
                )
            ),


            layers.Embedding(
                input_dim=vocab_size,
                output_dim=EMBEDDING_DIM,
                name="embedding"
            ),


            layers.LSTM(
                HIDDEN_SIZE,
                return_sequences=True,
                name="lstm_1"
            ),


            layers.Dropout(
                DROPOUT_RATE,
                name="dropout_1"
            ),


            layers.LSTM(
                HIDDEN_SIZE,
                name="lstm_2"
            ),


            layers.Dropout(
                DROPOUT_RATE,
                name="dropout_2"
            ),


            layers.Dense(
                vocab_size,
                activation="softmax",
                name="output"
            )
        ]
    )


    optimizer = optimizers.Adam(
        learning_rate=LEARNING_RATE,
        clipnorm=MAX_GRADIENT_NORM
    )


    model.compile(

        optimizer=optimizer,

        loss=(
            "sparse_categorical_crossentropy"
        ),

        metrics=[
            "accuracy"
        ]
    )


    return model


lstm_model = build_lstm_model(
    VOCAB_SIZE,
    SEQUENCE_LENGTH
)


print(
    "\n"
    + "=" * 75
)

print(
    "LSTM MODEL"
)

print(
    "=" * 75
)


lstm_model.summary()


# ======================================================================
# STEP 18: SAVE LSTM MODEL SUMMARY
#
# FIX FOR YOUR ERROR:
#
# UnicodeEncodeError:
# 'charmap' codec can't encode characters
#
# Keras model.summary() uses Unicode box characters.
# Windows may default to cp1252.
#
# encoding="utf-8" FIXES IT.
# ======================================================================

LSTM_SUMMARY_PATH = os.path.join(
    PROJECT_DIR,
    "lstm_model_summary.txt"
)


with open(
    LSTM_SUMMARY_PATH,
    "w",
    encoding="utf-8"
) as file:

    lstm_model.summary(
        print_fn=lambda line:
        file.write(
            line + "\n"
        )
    )


print(
    "\nLSTM model summary saved successfully."
)


# ======================================================================
# STEP 19: LIGHTWEIGHT CHECKPOINT CALLBACK
# ======================================================================

class LightweightCheckpoint(
    callbacks.Callback
):

    def __init__(
        self,
        directory
    ):

        super().__init__()

        self.directory = directory

        os.makedirs(
            directory,
            exist_ok=True
        )


    def on_epoch_end(
        self,
        epoch,
        logs=None
    ):

        current_epoch = (
            epoch + 1
        )


        if current_epoch % 5 == 0:

            path = os.path.join(
                self.directory,
                f"lstm_epoch_"
                f"{current_epoch:02d}"
                f".weights.h5"
            )


            self.model.save_weights(
                path
            )


            print(
                "\nCheckpoint saved:",
                path
            )


checkpoint_callback = (
    LightweightCheckpoint(
        CHECKPOINT_DIR
    )
)


# ======================================================================
# STEP 20: CSV LOGGER
# ======================================================================

csv_logger = callbacks.CSVLogger(
    os.path.join(
        PROJECT_DIR,
        "training_history.csv"
    ),
    append=False
)


# ======================================================================
# STEP 21: EARLY STOPPING
# ======================================================================

early_stopping = callbacks.EarlyStopping(

    monitor="val_loss",

    patience=8,

    restore_best_weights=True,

    verbose=1
)


# ======================================================================
# STEP 22: REDUCE LEARNING RATE
# ======================================================================

reduce_lr = callbacks.ReduceLROnPlateau(

    monitor="val_loss",

    factor=0.5,

    patience=3,

    min_lr=0.00001,

    verbose=1
)


# ======================================================================
# STEP 23: TEMPERATURE SAMPLING
# ======================================================================

def temperature_sampling(
    probabilities,
    temperature=1.0
):

    probabilities = np.asarray(
        probabilities,
        dtype=np.float64
    )


    probabilities = np.maximum(
        probabilities,
        1e-12
    )


    temperature = max(
        float(temperature),
        1e-6
    )


    logits = (
        np.log(
            probabilities
        )
        / temperature
    )


    logits -= np.max(
        logits
    )


    probabilities = np.exp(
        logits
    )


    probabilities /= np.sum(
        probabilities
    )


    return int(
        np.random.choice(
            len(probabilities),
            p=probabilities
        )
    )


# ======================================================================
# STEP 24: TOP-K SAMPLING
# ======================================================================

def top_k_sampling(
    probabilities,
    k=5,
    temperature=1.0
):

    probabilities = np.asarray(
        probabilities,
        dtype=np.float64
    )


    probabilities = np.maximum(
        probabilities,
        1e-12
    )


    temperature = max(
        float(temperature),
        1e-6
    )


    logits = (
        np.log(
            probabilities
        )
        / temperature
    )


    k = min(
        int(k),
        len(logits)
    )


    top_indices = np.argsort(
        logits
    )[-k:]


    top_logits = logits[
        top_indices
    ]


    top_logits -= np.max(
        top_logits
    )


    top_probabilities = np.exp(
        top_logits
    )


    top_probabilities /= np.sum(
        top_probabilities
    )


    return int(
        np.random.choice(
            top_indices,
            p=top_probabilities
        )
    )


# ======================================================================
# STEP 25: TOP-P SAMPLING
# ======================================================================

def top_p_sampling(
    probabilities,
    p=0.9,
    temperature=1.0
):

    probabilities = np.asarray(
        probabilities,
        dtype=np.float64
    )


    probabilities = np.maximum(
        probabilities,
        1e-12
    )


    temperature = max(
        float(temperature),
        1e-6
    )


    logits = (
        np.log(
            probabilities
        )
        / temperature
    )


    logits -= np.max(
        logits
    )


    probabilities = np.exp(
        logits
    )


    probabilities /= np.sum(
        probabilities
    )


    sorted_indices = np.argsort(
        probabilities
    )[::-1]


    sorted_probabilities = probabilities[
        sorted_indices
    ]


    cumulative_probabilities = np.cumsum(
        sorted_probabilities
    )


    cutoff = (
        np.searchsorted(
            cumulative_probabilities,
            p
        )
        + 1
    )


    cutoff = max(
        cutoff,
        1
    )


    selected_indices = sorted_indices[
        :cutoff
    ]


    selected_probabilities = probabilities[
        selected_indices
    ]


    selected_probabilities /= np.sum(
        selected_probabilities
    )


    return int(
        np.random.choice(
            selected_indices,
            p=selected_probabilities
        )
    )


# ======================================================================
# STEP 26: TEXT GENERATION FUNCTION
# ======================================================================

def generate_text(
    model,
    seed_text,
    length=500,
    temperature=1.0,
    strategy="temperature",
    top_k=5,
    top_p=0.9,
    sequence_length=SEQUENCE_LENGTH
):

    generated_text = seed_text


    default_index = char2idx.get(
        " ",
        0
    )


    encoded_seed = [
        char2idx.get(
            character,
            default_index
        )
        for character
        in seed_text
    ]


    # --------------------------------------------------------------
    # Pad short seed
    # --------------------------------------------------------------

    if len(encoded_seed) < sequence_length:

        padding = [
            default_index
        ] * (
            sequence_length
            - len(encoded_seed)
        )


        encoded_seed = (
            padding
            + encoded_seed
        )


    # --------------------------------------------------------------
    # Keep latest characters only
    # --------------------------------------------------------------

    encoded_seed = encoded_seed[
        -sequence_length:
    ]


    # --------------------------------------------------------------
    # Generate
    # --------------------------------------------------------------

    for _ in range(
        length
    ):

        input_sequence = np.array(
            [
                encoded_seed
            ],
            dtype=np.int32
        )


        probabilities = model.predict(
            input_sequence,
            verbose=0
        )[0]


        if strategy == "greedy":

            next_index = int(
                np.argmax(
                    probabilities
                )
            )


        elif strategy == "top_k":

            next_index = top_k_sampling(
                probabilities,
                k=top_k,
                temperature=temperature
            )


        elif strategy == "top_p":

            next_index = top_p_sampling(
                probabilities,
                p=top_p,
                temperature=temperature
            )


        else:

            next_index = temperature_sampling(
                probabilities,
                temperature=temperature
            )


        next_character = idx2char[
            next_index
        ]


        generated_text += (
            next_character
        )


        encoded_seed.append(
            next_index
        )


        encoded_seed = encoded_seed[
            -sequence_length:
        ]


    return generated_text


# ======================================================================
# STEP 27: SAMPLE GENERATION CALLBACK
# ======================================================================

class TextGenerationCallback(
    callbacks.Callback
):

    def __init__(
        self,
        seed_text
    ):

        super().__init__()

        self.seed_text = (
            seed_text
        )


    def on_epoch_end(
        self,
        epoch,
        logs=None
    ):

        current_epoch = (
            epoch + 1
        )


        if current_epoch % 5 != 0:

            return


        print(
            "\n"
            + "=" * 70
        )

        print(
            f"SAMPLE AFTER EPOCH "
            f"{current_epoch}"
        )

        print(
            "=" * 70
        )


        generated = generate_text(

            self.model,

            self.seed_text,

            length=300,

            temperature=0.8,

            strategy="temperature"
        )


        print(
            generated[:500]
        )


        sample_file = os.path.join(
            SAMPLES_DIR,
            f"epoch_"
            f"{current_epoch:02d}"
            f"_sample.txt"
        )


        with open(
            sample_file,
            "w",
            encoding="utf-8"
        ) as file:

            file.write(
                generated
            )


# ======================================================================
# STEP 28: CALLBACK SEED
# ======================================================================

random_start = random.randint(
    0,
    len(text)
    - SEQUENCE_LENGTH
    - 1
)


callback_seed = text[
    random_start:
    random_start
    + SEQUENCE_LENGTH
]


generation_callback = (
    TextGenerationCallback(
        callback_seed
    )
)


# ======================================================================
# STEP 29: TRAIN LSTM
# ======================================================================

print(
    "\n"
    + "=" * 75
)

print(
    "STARTING LSTM TRAINING"
)

print(
    "=" * 75
)


history = lstm_model.fit(

    train_dataset,

    validation_data=validation_dataset,

    epochs=EPOCHS,

    callbacks=[

        checkpoint_callback,

        csv_logger,

        early_stopping,

        reduce_lr,

        generation_callback

    ],

    verbose=1
)


# ======================================================================
# STEP 30: SAVE FINAL LSTM WEIGHTS
# ======================================================================

FINAL_LSTM_WEIGHTS = os.path.join(
    PROJECT_DIR,
    "lstm_final.weights.h5"
)


lstm_model.save_weights(
    FINAL_LSTM_WEIGHTS
)


print(
    "\nFinal LSTM weights saved."
)


# ======================================================================
# STEP 31: SAVE FULL LSTM MODEL
# ======================================================================

FINAL_LSTM_MODEL = os.path.join(
    PROJECT_DIR,
    "character_lstm_model.h5"
)


lstm_model.save(
    FINAL_LSTM_MODEL,
    include_optimizer=False
)


print(
    "Final LSTM H5 model saved."
)


# ======================================================================
# STEP 32: SAVE TRAINING HISTORY JSON
# ======================================================================

history_json = {

    key: [
        float(value)
        for value
        in values
    ]

    for key, values
    in history.history.items()
}


with open(
    os.path.join(
        PROJECT_DIR,
        "training_history.json"
    ),
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        history_json,
        file,
        indent=4,
        ensure_ascii=False
    )


# ======================================================================
# STEP 33: LOSS GRAPH
# ======================================================================

epochs_range = range(
    1,
    len(
        history.history[
            "loss"
        ]
    ) + 1
)


plt.figure(
    figsize=(10, 6)
)


plt.plot(
    epochs_range,
    history.history[
        "loss"
    ],
    marker="o",
    label="Training Loss"
)


plt.plot(
    epochs_range,
    history.history[
        "val_loss"
    ],
    marker="o",
    label="Validation Loss"
)


plt.title(
    "LSTM Training and Validation Loss"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Loss"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()


plt.savefig(
    os.path.join(
        PROJECT_DIR,
        "training_validation_loss.png"
    ),
    dpi=150,
    bbox_inches="tight"
)


plt.show()

plt.close()


# ======================================================================
# STEP 34: ACCURACY GRAPH
# ======================================================================

plt.figure(
    figsize=(10, 6)
)


plt.plot(
    epochs_range,
    history.history[
        "accuracy"
    ],
    marker="o",
    label="Training Accuracy"
)


plt.plot(
    epochs_range,
    history.history[
        "val_accuracy"
    ],
    marker="o",
    label="Validation Accuracy"
)


plt.title(
    "LSTM Training and Validation Accuracy"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()


plt.savefig(
    os.path.join(
        PROJECT_DIR,
        "accuracy_graph.png"
    ),
    dpi=150,
    bbox_inches="tight"
)


plt.show()

plt.close()


# ======================================================================
# STEP 35: PERPLEXITY
#
# Perplexity = exp(loss)
# ======================================================================

training_perplexity = [

    math.exp(
        min(
            float(loss),
            20
        )
    )

    for loss
    in history.history[
        "loss"
    ]
]


validation_perplexity = [

    math.exp(
        min(
            float(loss),
            20
        )
    )

    for loss
    in history.history[
        "val_loss"
    ]
]


perplexity_df = pd.DataFrame(
    {

        "Epoch":
            list(
                epochs_range
            ),

        "Training_Perplexity":
            training_perplexity,

        "Validation_Perplexity":
            validation_perplexity
    }
)


perplexity_df.to_csv(
    os.path.join(
        RESULTS_DIR,
        "perplexity_results.csv"
    ),
    index=False,
    encoding="utf-8"
)


# ======================================================================
# STEP 36: PERPLEXITY GRAPH
# ======================================================================

plt.figure(
    figsize=(10, 6)
)


plt.plot(
    epochs_range,
    training_perplexity,
    marker="o",
    label="Training Perplexity"
)


plt.plot(
    epochs_range,
    validation_perplexity,
    marker="o",
    label="Validation Perplexity"
)


plt.title(
    "Perplexity Over Epochs"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Perplexity"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()


plt.savefig(
    os.path.join(
        PROJECT_DIR,
        "perplexity_graph.png"
    ),
    dpi=150,
    bbox_inches="tight"
)


plt.show()

plt.close()


# ======================================================================
# STEP 37: EVALUATE LSTM
# ======================================================================

lstm_val_loss, lstm_val_accuracy = (
    lstm_model.evaluate(
        validation_dataset,
        verbose=1
    )
)


lstm_perplexity = math.exp(
    min(
        float(
            lstm_val_loss
        ),
        20
    )
)


print(
    "\n"
    + "=" * 75
)

print(
    "FINAL LSTM RESULTS"
)

print(
    "=" * 75
)


print(
    f"Validation Loss     : "
    f"{lstm_val_loss:.4f}"
)


print(
    f"Validation Accuracy : "
    f"{lstm_val_accuracy:.4f}"
)


print(
    f"Perplexity          : "
    f"{lstm_perplexity:.4f}"
)


# ======================================================================
# STEP 38: BUILD GRU MODEL
# ======================================================================

def build_gru_model(
    vocab_size,
    sequence_length=100
):

    model = models.Sequential(
        [

            layers.Input(
                shape=(
                    sequence_length,
                )
            ),


            layers.Embedding(
                input_dim=vocab_size,
                output_dim=EMBEDDING_DIM,
                name="gru_embedding"
            ),


            layers.GRU(
                HIDDEN_SIZE,
                return_sequences=True,
                name="gru_1"
            ),


            layers.Dropout(
                DROPOUT_RATE,
                name="gru_dropout_1"
            ),


            layers.GRU(
                HIDDEN_SIZE,
                name="gru_2"
            ),


            layers.Dropout(
                DROPOUT_RATE,
                name="gru_dropout_2"
            ),


            layers.Dense(
                vocab_size,
                activation="softmax",
                name="gru_output"
            )
        ]
    )


    optimizer = optimizers.Adam(
        learning_rate=LEARNING_RATE,
        clipnorm=MAX_GRADIENT_NORM
    )


    model.compile(

        optimizer=optimizer,

        loss=(
            "sparse_categorical_crossentropy"
        ),

        metrics=[
            "accuracy"
        ]
    )


    return model


gru_model = build_gru_model(
    VOCAB_SIZE,
    SEQUENCE_LENGTH
)


print(
    "\n"
    + "=" * 75
)

print(
    "GRU MODEL"
)

print(
    "=" * 75
)


gru_model.summary()


# ======================================================================
# STEP 39: SAVE GRU MODEL SUMMARY
#
# UTF-8 FIX ALSO APPLIED HERE
# ======================================================================

GRU_SUMMARY_PATH = os.path.join(
    PROJECT_DIR,
    "gru_model_summary.txt"
)


with open(
    GRU_SUMMARY_PATH,
    "w",
    encoding="utf-8"
) as file:

    gru_model.summary(
        print_fn=lambda line:
        file.write(
            line + "\n"
        )
    )


print(
    "\nGRU model summary saved successfully."
)


# ======================================================================
# STEP 40: TRAIN GRU
#
# 20 epochs maximum to reduce training time.
# ======================================================================

GRU_EPOCHS = min(
    EPOCHS,
    20
)


gru_history = gru_model.fit(

    train_dataset,

    validation_data=validation_dataset,

    epochs=GRU_EPOCHS,

    callbacks=[

        callbacks.EarlyStopping(

            monitor="val_loss",

            patience=5,

            restore_best_weights=True,

            verbose=1
        ),


        callbacks.ReduceLROnPlateau(

            monitor="val_loss",

            factor=0.5,

            patience=3,

            min_lr=0.00001,

            verbose=1
        )

    ],

    verbose=1
)


# ======================================================================
# STEP 41: EVALUATE GRU
# ======================================================================

gru_val_loss, gru_val_accuracy = (
    gru_model.evaluate(
        validation_dataset,
        verbose=1
    )
)


gru_perplexity = math.exp(
    min(
        float(
            gru_val_loss
        ),
        20
    )
)


print(
    "\n"
    + "=" * 75
)

print(
    "FINAL GRU RESULTS"
)

print(
    "=" * 75
)


print(
    f"Validation Loss     : "
    f"{gru_val_loss:.4f}"
)


print(
    f"Validation Accuracy : "
    f"{gru_val_accuracy:.4f}"
)


print(
    f"Perplexity          : "
    f"{gru_perplexity:.4f}"
)


# ======================================================================
# STEP 42: SAVE GRU WEIGHTS
# ======================================================================

GRU_WEIGHTS = os.path.join(
    PROJECT_DIR,
    "gru_final.weights.h5"
)


gru_model.save_weights(
    GRU_WEIGHTS
)


# ======================================================================
# STEP 43: LSTM VS GRU COMPARISON
# ======================================================================

comparison_df = pd.DataFrame(
    {

        "Model": [
            "LSTM",
            "GRU"
        ],


        "Validation_Loss": [
            float(
                lstm_val_loss
            ),
            float(
                gru_val_loss
            )
        ],


        "Validation_Accuracy": [
            float(
                lstm_val_accuracy
            ),
            float(
                gru_val_accuracy
            )
        ],


        "Perplexity": [
            float(
                lstm_perplexity
            ),
            float(
                gru_perplexity
            )
        ],


        "Parameters": [
            int(
                lstm_model.count_params()
            ),
            int(
                gru_model.count_params()
            )
        ]
    }
)


comparison_df.to_csv(
    os.path.join(
        RESULTS_DIR,
        "lstm_vs_gru_comparison.csv"
    ),
    index=False,
    encoding="utf-8"
)


print(
    "\nLSTM vs GRU:"
)

print(
    comparison_df.to_string(
        index=False
    )
)


# ======================================================================
# STEP 44: LSTM VS GRU GRAPH
# ======================================================================

plt.figure(
    figsize=(9, 6)
)


plt.bar(
    comparison_df[
        "Model"
    ],
    comparison_df[
        "Validation_Loss"
    ]
)


plt.title(
    "LSTM vs GRU Validation Loss"
)

plt.xlabel(
    "Model"
)

plt.ylabel(
    "Validation Loss"
)

plt.tight_layout()


plt.savefig(
    os.path.join(
        PROJECT_DIR,
        "lstm_gru_comparison.png"
    ),
    dpi=150,
    bbox_inches="tight"
)


plt.show()

plt.close()


# ======================================================================
# STEP 45: SEQUENCE LENGTH EXPERIMENT
#
# Compare:
# 50
# 100
# 200
#
# Uses smaller models and only 3 epochs to keep experiment practical.
# ======================================================================

def sequence_length_experiment():

    sequence_lengths = [
        50,
        100,
        200
    ]


    experiment_results = []


    # Use subset to reduce training time
    experiment_text = encoded_text[
        :min(
            len(encoded_text),
            250000
        )
    ]


    for seq_len in sequence_lengths:

        print(
            "\n"
            + "=" * 60
        )

        print(
            "Testing sequence length:",
            seq_len
        )

        print(
            "=" * 60
        )


        exp_X, exp_y = create_sequences(

            experiment_text,

            sequence_length=seq_len,

            stride=5
        )


        split = int(
            0.90
            * len(exp_X)
        )


        exp_X_train = exp_X[
            :split
        ]

        exp_y_train = exp_y[
            :split
        ]

        exp_X_val = exp_X[
            split:
        ]

        exp_y_val = exp_y[
            split:
        ]


        exp_model = models.Sequential(
            [

                layers.Input(
                    shape=(
                        seq_len,
                    )
                ),


                layers.Embedding(
                    VOCAB_SIZE,
                    64
                ),


                layers.LSTM(
                    128
                ),


                layers.Dense(
                    VOCAB_SIZE,
                    activation="softmax"
                )
            ]
        )


        exp_model.compile(

            optimizer=optimizers.Adam(
                learning_rate=0.001,
                clipnorm=5.0
            ),

            loss=(
                "sparse_categorical_crossentropy"
            ),

            metrics=[
                "accuracy"
            ]
        )


        exp_history = exp_model.fit(

            exp_X_train,

            exp_y_train,

            validation_data=(
                exp_X_val,
                exp_y_val
            ),

            batch_size=128,

            epochs=3,

            verbose=1
        )


        final_loss = float(
            exp_history.history[
                "val_loss"
            ][-1]
        )


        final_accuracy = float(
            exp_history.history[
                "val_accuracy"
            ][-1]
        )


        experiment_results.append(
            {

                "Sequence_Length":
                    seq_len,

                "Validation_Loss":
                    final_loss,

                "Validation_Accuracy":
                    final_accuracy,

                "Perplexity":
                    math.exp(
                        min(
                            final_loss,
                            20
                        )
                    ),

                "Training_Samples":
                    len(
                        exp_X_train
                    )
            }
        )


        del exp_model


    return pd.DataFrame(
        experiment_results
    )


sequence_results = (
    sequence_length_experiment()
)


sequence_results.to_csv(
    os.path.join(
        RESULTS_DIR,
        "sequence_length_experiment.csv"
    ),
    index=False,
    encoding="utf-8"
)


print(
    "\nSequence Length Results:"
)

print(
    sequence_results.to_string(
        index=False
    )
)


# ======================================================================
# STEP 46: SEQUENCE LENGTH GRAPH
# ======================================================================

plt.figure(
    figsize=(9, 6)
)


plt.plot(
    sequence_results[
        "Sequence_Length"
    ],
    sequence_results[
        "Validation_Loss"
    ],
    marker="o"
)


plt.title(
    "Effect of Sequence Length on Validation Loss"
)

plt.xlabel(
    "Sequence Length"
)

plt.ylabel(
    "Validation Loss"
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()


plt.savefig(
    os.path.join(
        PROJECT_DIR,
        "sequence_length_comparison.png"
    ),
    dpi=150,
    bbox_inches="tight"
)


plt.show()

plt.close()


# ======================================================================
# STEP 47: CREATE RANDOM SEEDS
# ======================================================================

seed_texts = []


for _ in range(
    10
):

    start = random.randint(
        0,
        len(text)
        - SEQUENCE_LENGTH
        - 1
    )


    seed = text[
        start:
        start
        + SEQUENCE_LENGTH
    ]


    seed_texts.append(
        seed
    )


with open(
    os.path.join(
        RESULTS_DIR,
        "seed_texts.txt"
    ),
    "w",
    encoding="utf-8"
) as file:

    for number, seed in enumerate(
        seed_texts,
        start=1
    ):

        file.write(
            f"SEED {number}\n"
        )

        file.write(
            "=" * 60
            + "\n"
        )

        file.write(
            seed
        )

        file.write(
            "\n\n"
        )


# ======================================================================
# STEP 48: GENERATE REQUIRED 30 TEXT SAMPLES
#
# 10 at 0.5
# 10 at 1.0
# 10 at 1.2
# ======================================================================

required_temperatures = [
    0.5,
    1.0,
    1.2
]


ALL_SAMPLES_FILE = os.path.join(
    PROJECT_DIR,
    "generated_text_samples.txt"
)


with open(
    ALL_SAMPLES_FILE,
    "w",
    encoding="utf-8"
) as output_file:


    for temperature in required_temperatures:


        print(
            "\nGenerating 10 samples at temperature:",
            temperature
        )


        for sample_number in range(
            10
        ):


            seed = seed_texts[
                sample_number
            ]


            generated = generate_text(

                lstm_model,

                seed,

                length=GENERATE_LENGTH,

                temperature=temperature,

                strategy="temperature"
            )


            header = (

                "\n"
                + "=" * 75
                + "\n"

                + f"TEMPERATURE: "
                + f"{temperature}"

                + f" | SAMPLE: "
                + f"{sample_number + 1}"

                + "\n"
                + "=" * 75
                + "\n"
            )


            output_file.write(
                header
            )


            output_file.write(
                generated
            )


            output_file.write(
                "\n\n"
            )


            individual_file = os.path.join(

                SAMPLES_DIR,

                f"temperature_"
                f"{temperature}_"
                f"sample_"
                f"{sample_number + 1}"
                f".txt"
            )


            with open(
                individual_file,
                "w",
                encoding="utf-8"
            ) as sample_file:

                sample_file.write(
                    generated
                )


print(
    "\n30 required samples generated successfully."
)


# ======================================================================
# STEP 49: TEMPERATURE 0.2 SAMPLE
# ======================================================================

temperature_02_sample = generate_text(

    lstm_model,

    seed_texts[0],

    length=500,

    temperature=0.2,

    strategy="temperature"
)


with open(
    os.path.join(
        SAMPLES_DIR,
        "temperature_0.2_sample.txt"
    ),
    "w",
    encoding="utf-8"
) as file:

    file.write(
        temperature_02_sample
    )


# ======================================================================
# STEP 50: SAMPLING STRATEGY COMPARISON
# ======================================================================

strategy_samples = {}


strategy_samples[
    "Greedy"
] = generate_text(

    lstm_model,

    seed_texts[0],

    length=500,

    strategy="greedy"
)


strategy_samples[
    "Temperature 1.0"
] = generate_text(

    lstm_model,

    seed_texts[0],

    length=500,

    temperature=1.0,

    strategy="temperature"
)


strategy_samples[
    "Top-k 5"
] = generate_text(

    lstm_model,

    seed_texts[0],

    length=500,

    temperature=1.0,

    strategy="top_k",

    top_k=5
)


strategy_samples[
    "Top-k 10"
] = generate_text(

    lstm_model,

    seed_texts[0],

    length=500,

    temperature=1.0,

    strategy="top_k",

    top_k=10
)


strategy_samples[
    "Top-p 0.9"
] = generate_text(

    lstm_model,

    seed_texts[0],

    length=500,

    temperature=1.0,

    strategy="top_p",

    top_p=0.9
)


with open(
    os.path.join(
        PROJECT_DIR,
        "sampling_strategy_comparison.txt"
    ),
    "w",
    encoding="utf-8"
) as file:


    for strategy, sample in (
        strategy_samples.items()
    ):


        file.write(
            "\n"
            + "=" * 75
            + "\n"
        )


        file.write(
            strategy
            + "\n"
        )


        file.write(
            "=" * 75
            + "\n"
        )


        file.write(
            sample
        )


        file.write(
            "\n\n"
        )


# ======================================================================
# STEP 51: TEMPERATURE DIVERSITY ANALYSIS
# ======================================================================

def calculate_diversity(
    generated_text
):

    if not generated_text:

        return 0.0, 0.0


    char_diversity = (

        len(
            set(
                generated_text
            )
        )

        / len(
            generated_text
        )
    )


    bigrams = [

        generated_text[
            i:i + 2
        ]

        for i in range(
            len(
                generated_text
            ) - 1
        )
    ]


    if len(
        bigrams
    ) > 0:

        bigram_diversity = (

            len(
                set(
                    bigrams
                )
            )

            / len(
                bigrams
            )
        )

    else:

        bigram_diversity = 0.0


    return (
        char_diversity,
        bigram_diversity
    )


temperature_results = []


temperatures = [
    0.2,
    0.5,
    1.0,
    1.2
]


for temperature in temperatures:


    sample = generate_text(

        lstm_model,

        seed_texts[1],

        length=1000,

        temperature=temperature,

        strategy="temperature"
    )


    char_diversity, bigram_diversity = (
        calculate_diversity(
            sample
        )
    )


    # Descriptive proxy, not a formal linguistic coherence metric
    coherence_proxy = (
        1.0
        - bigram_diversity
    )


    temperature_results.append(
        {

            "Temperature":
                temperature,

            "Character_Diversity":
                char_diversity,

            "Bigram_Diversity":
                bigram_diversity,

            "Coherence_Proxy":
                coherence_proxy
        }
    )


temperature_df = pd.DataFrame(
    temperature_results
)


temperature_df.to_csv(
    os.path.join(
        RESULTS_DIR,
        "temperature_analysis.csv"
    ),
    index=False,
    encoding="utf-8"
)


# ======================================================================
# STEP 52: TEMPERATURE GRAPH
# ======================================================================

plt.figure(
    figsize=(10, 6)
)


plt.plot(
    temperature_df[
        "Temperature"
    ],
    temperature_df[
        "Bigram_Diversity"
    ],
    marker="o",
    label="Diversity"
)


plt.plot(
    temperature_df[
        "Temperature"
    ],
    temperature_df[
        "Coherence_Proxy"
    ],
    marker="o",
    label="Coherence Proxy"
)


plt.title(
    "Temperature Effect: Diversity vs Coherence"
)

plt.xlabel(
    "Temperature"
)

plt.ylabel(
    "Score"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()


plt.savefig(
    os.path.join(
        PROJECT_DIR,
        "temperature_effect_comparison.png"
    ),
    dpi=150,
    bbox_inches="tight"
)


plt.show()

plt.close()


# ======================================================================
# STEP 53: ORIGINAL VS GENERATED CHARACTER FREQUENCY
# ======================================================================

generated_for_frequency = generate_text(

    lstm_model,

    seed_texts[2],

    length=5000,

    temperature=0.8,

    strategy="temperature"
)


original_counter = Counter(
    text
)


generated_counter = Counter(
    generated_for_frequency
)


frequency_comparison = []


for character in characters:


    original_frequency = (

        original_counter[
            character
        ]

        / len(text)
    )


    generated_frequency = (

        generated_counter[
            character
        ]

        / len(
            generated_for_frequency
        )
    )


    frequency_comparison.append(
        {

            "Character":
                repr(
                    character
                ),

            "Original_Frequency":
                original_frequency,

            "Generated_Frequency":
                generated_frequency
        }
    )


frequency_comparison_df = pd.DataFrame(
    frequency_comparison
)


frequency_comparison_df.to_csv(
    os.path.join(
        RESULTS_DIR,
        "original_vs_generated_frequency.csv"
    ),
    index=False,
    encoding="utf-8"
)


# ======================================================================
# STEP 54: FREQUENCY COMPARISON GRAPH
# ======================================================================

top_characters = [

    character

    for character, count
    in original_counter.most_common(
        25
    )
]


original_values = [

    original_counter[
        character
    ]
    / len(text)

    for character
    in top_characters
]


generated_values = [

    generated_counter[
        character
    ]
    / len(
        generated_for_frequency
    )

    for character
    in top_characters
]


x_positions = np.arange(
    len(
        top_characters
    )
)


bar_width = 0.4


plt.figure(
    figsize=(14, 7)
)


plt.bar(
    x_positions
    - bar_width / 2,
    original_values,
    width=bar_width,
    label="Original"
)


plt.bar(
    x_positions
    + bar_width / 2,
    generated_values,
    width=bar_width,
    label="Generated"
)


plt.xticks(
    x_positions,
    [
        repr(
            character
        )
        for character
        in top_characters
    ],
    rotation=90
)


plt.title(
    "Character Frequency: Original vs Generated"
)

plt.xlabel(
    "Character"
)

plt.ylabel(
    "Relative Frequency"
)

plt.legend()

plt.tight_layout()


plt.savefig(
    os.path.join(
        PROJECT_DIR,
        "original_vs_generated_frequency.png"
    ),
    dpi=150,
    bbox_inches="tight"
)


plt.show()

plt.close()


# ======================================================================
# STEP 55: HIDDEN STATE MODEL
#
# Extract activations from second LSTM layer
# ======================================================================

hidden_model = tf.keras.Model(

    inputs=lstm_model.inputs,

    outputs=lstm_model.get_layer(
        "lstm_2"
    ).output
)


# ======================================================================
# STEP 56: SELECT t-SNE SAMPLES
# ======================================================================

TSNE_SAMPLES = min(
    1000,
    len(X_val)
)


sample_indices = np.random.choice(

    len(X_val),

    TSNE_SAMPLES,

    replace=False
)


tsne_inputs = X_val[
    sample_indices
]


tsne_targets = y_val[
    sample_indices
]


hidden_states = hidden_model.predict(

    tsne_inputs,

    batch_size=128,

    verbose=1
)


print(
    "\nHidden state shape:",
    hidden_states.shape
)


# ======================================================================
# STEP 57: t-SNE
# ======================================================================

tsne = TSNE(

    n_components=2,

    random_state=SEED,

    perplexity=30,

    init="pca",

    learning_rate="auto"
)


hidden_2d = tsne.fit_transform(
    hidden_states
)


# ======================================================================
# STEP 58: SAVE t-SNE RESULTS
# ======================================================================

tsne_df = pd.DataFrame(
    {

        "TSNE_1":
            hidden_2d[
                :, 0
            ],

        "TSNE_2":
            hidden_2d[
                :, 1
            ],

        "Target_Index":
            tsne_targets,

        "Target_Character": [

            repr(
                idx2char[
                    int(index)
                ]
            )

            for index
            in tsne_targets
        ]
    }
)


tsne_df.to_csv(
    os.path.join(
        RESULTS_DIR,
        "hidden_state_tsne.csv"
    ),
    index=False,
    encoding="utf-8"
)


# ======================================================================
# STEP 59: t-SNE GRAPH
# ======================================================================

plt.figure(
    figsize=(10, 8)
)


scatter = plt.scatter(

    hidden_2d[
        :, 0
    ],

    hidden_2d[
        :, 1
    ],

    c=tsne_targets,

    s=12,

    alpha=0.7
)


plt.title(
    "t-SNE Visualization of LSTM Hidden States"
)

plt.xlabel(
    "t-SNE Dimension 1"
)

plt.ylabel(
    "t-SNE Dimension 2"
)

plt.colorbar(
    scatter,
    label="Next Character Index"
)

plt.tight_layout()


plt.savefig(
    os.path.join(
        PROJECT_DIR,
        "hidden_state_tsne.png"
    ),
    dpi=150,
    bbox_inches="tight"
)


plt.show()

plt.close()


# ======================================================================
# STEP 60: CHARACTER BIGRAM ANALYSIS
# ======================================================================

def get_ngrams(
    input_text,
    n=2
):

    return [

        input_text[
            i:i+n
        ]

        for i in range(
            len(input_text)
            - n
            + 1
        )
    ]


original_bigrams = Counter(
    get_ngrams(
        text,
        2
    )
)


generated_bigrams = Counter(
    get_ngrams(
        generated_for_frequency,
        2
    )
)


top_original_bigrams = (
    original_bigrams.most_common(
        30
    )
)


bigram_rows = []


for bigram, original_count in (
    top_original_bigrams
):

    generated_count = (
        generated_bigrams[
            bigram
        ]
    )


    bigram_rows.append(
        {

            "Bigram":
                repr(
                    bigram
                ),

            "Original_Count":
                original_count,

            "Generated_Count":
                generated_count
        }
    )


bigram_df = pd.DataFrame(
    bigram_rows
)


bigram_df.to_csv(
    os.path.join(
        RESULTS_DIR,
        "character_bigram_analysis.csv"
    ),
    index=False,
    encoding="utf-8"
)


# ======================================================================
# STEP 61: BIGRAM GRAPH
# ======================================================================

positions = np.arange(
    len(
        bigram_df
    )
)


width = 0.4


original_normalized = (

    bigram_df[
        "Original_Count"
    ]

    / max(
        bigram_df[
            "Original_Count"
        ].max(),
        1
    )
)


generated_normalized = (

    bigram_df[
        "Generated_Count"
    ]

    / max(
        bigram_df[
            "Generated_Count"
        ].max(),
        1
    )
)


plt.figure(
    figsize=(14, 7)
)


plt.bar(
    positions
    - width / 2,
    original_normalized,
    width,
    label="Original"
)


plt.bar(
    positions
    + width / 2,
    generated_normalized,
    width,
    label="Generated"
)


plt.xticks(
    positions,
    bigram_df[
        "Bigram"
    ],
    rotation=90
)


plt.title(
    "Character Bigram Comparison"
)

plt.xlabel(
    "Character Bigram"
)

plt.ylabel(
    "Normalized Frequency"
)

plt.legend()

plt.tight_layout()


plt.savefig(
    os.path.join(
        PROJECT_DIR,
        "character_bigram_comparison.png"
    ),
    dpi=150,
    bbox_inches="tight"
)


plt.show()

plt.close()


# ======================================================================
# STEP 62: SAVE FINAL METRICS
# ======================================================================

final_metrics = {

    "LSTM": {

        "validation_loss":
            float(
                lstm_val_loss
            ),

        "validation_accuracy":
            float(
                lstm_val_accuracy
            ),

        "perplexity":
            float(
                lstm_perplexity
            ),

        "parameters":
            int(
                lstm_model.count_params()
            )
    },


    "GRU": {

        "validation_loss":
            float(
                gru_val_loss
            ),

        "validation_accuracy":
            float(
                gru_val_accuracy
            ),

        "perplexity":
            float(
                gru_perplexity
            ),

        "parameters":
            int(
                gru_model.count_params()
            )
    }
}


with open(
    os.path.join(
        PROJECT_DIR,
        "final_metrics.json"
    ),
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        final_metrics,
        file,
        indent=4,
        ensure_ascii=False
    )


# ======================================================================
# STEP 63: SAVE CONFIGURATION
# ======================================================================

configuration = {

    "project":
        "Character-Level RNN Text Generation",

    "dataset":
        "Shakespeare Complete Works",

    "sequence_length":
        SEQUENCE_LENGTH,

    "stride":
        STRIDE,

    "embedding_dimension":
        EMBEDDING_DIM,

    "hidden_size":
        HIDDEN_SIZE,

    "dropout":
        DROPOUT_RATE,

    "batch_size":
        BATCH_SIZE,

    "epochs":
        EPOCHS,

    "optimizer":
        "Adam",

    "learning_rate":
        LEARNING_RATE,

    "gradient_clipping":
        MAX_GRADIENT_NORM,

    "train_split":
        0.90,

    "validation_split":
        0.10,

    "sampling_methods": [
        "greedy",
        "temperature",
        "top-k",
        "top-p"
    ],

    "temperatures": [
        0.2,
        0.5,
        1.0,
        1.2
    ]
}


with open(
    os.path.join(
        PROJECT_DIR,
        "config.json"
    ),
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        configuration,
        file,
        indent=4,
        ensure_ascii=False
    )


# ======================================================================
# STEP 64: INTERACTIVE GENERATION FUNCTION
#
# Run:
#
# interactive_generation()
#
# after training if required.
# ======================================================================

def interactive_generation():

    print(
        "\n"
        + "=" * 75
    )

    print(
        "INTERACTIVE TEXT GENERATION"
    )

    print(
        "=" * 75
    )


    user_seed = input(
        "\nEnter seed text: "
    )


    try:

        generation_length = int(
            input(
                "Generation length: "
            )
        )

    except:

        generation_length = 500


    try:

        temperature = float(
            input(
                "Temperature "
                "(0.2 / 0.5 / 1.0 / 1.2): "
            )
        )

    except:

        temperature = 0.8


    generated = generate_text(

        lstm_model,

        user_seed,

        length=generation_length,

        temperature=temperature,

        strategy="temperature"
    )


    print(
        "\nGENERATED TEXT:"
    )

    print(
        generated
    )


    with open(
        os.path.join(
            SAMPLES_DIR,
            "interactive_generation.txt"
        ),
        "w",
        encoding="utf-8"
    ) as file:

        file.write(
            generated
        )


    return generated


# ======================================================================
# STEP 65: ANALYSIS REPORT
# ======================================================================

analysis_text = f"""
CHARACTER-LEVEL RECURRENT NEURAL NETWORK FOR TEXT GENERATION
=============================================================

PROJECT
-------
Character-Level RNN for Text Generation using Shakespeare Corpus


DATASET
-------
Dataset: Shakespeare Complete Works

Total Characters:
{len(text)}

Vocabulary Size:
{VOCAB_SIZE}

Sequence Length:
{SEQUENCE_LENGTH}

Sliding Window Stride:
{STRIDE}

Training Split:
90%

Validation Split:
10%


LSTM ARCHITECTURE
-----------------
Embedding Dimension:
{EMBEDDING_DIM}

First LSTM:
{HIDDEN_SIZE} units with return_sequences=True

Dropout:
{DROPOUT_RATE}

Second LSTM:
{HIDDEN_SIZE} units

Output:
Dense layer with {VOCAB_SIZE} classes and Softmax activation


TRAINING
--------
Optimizer:
Adam

Learning Rate:
{LEARNING_RATE}

Batch Size:
{BATCH_SIZE}

Maximum Epochs:
{EPOCHS}

Gradient Clipping:
{MAX_GRADIENT_NORM}


LSTM RESULTS
------------
Validation Loss:
{lstm_val_loss:.4f}

Validation Accuracy:
{lstm_val_accuracy:.4f}

Perplexity:
{lstm_perplexity:.4f}


GRU RESULTS
-----------
Validation Loss:
{gru_val_loss:.4f}

Validation Accuracy:
{gru_val_accuracy:.4f}

Perplexity:
{gru_perplexity:.4f}


TEMPERATURE EFFECT
------------------
Temperature 0.2 produces conservative and predictable text.

Temperature 0.5 generally improves coherence while maintaining some
variation.

Temperature 1.0 provides balanced sampling.

Temperature 1.2 increases creativity and diversity but can also increase
spelling and grammatical errors.


SAMPLING METHODS
----------------
Greedy sampling always selects the highest probability character.

Temperature sampling changes the sharpness of the predicted probability
distribution.

Top-k sampling restricts character selection to the k most probable
characters.

Top-p or nucleus sampling selects from the smallest group of characters
whose cumulative probability reaches p.


LSTM VS GRU
-----------
Both LSTM and GRU use gating mechanisms to address problems associated
with learning longer dependencies in vanilla recurrent networks.

GRU generally contains fewer parameters because its gating architecture
is simpler.

LSTM maintains both hidden state and cell state, allowing more explicit
control over long-term information.


CHARACTER PATTERNS
------------------
The trained character-level model learns several characteristics of the
training corpus, including:

1. Character combinations
2. Word boundaries
3. Capitalization
4. Punctuation
5. Newline structure
6. Common words
7. Shakespeare-like dialogue formatting


SEQUENCE LENGTH
---------------
A shorter sequence reduces computation and training time but gives the
model less historical context.

A longer sequence provides more context but increases memory usage and
training time.

Sequence lengths 50, 100 and 200 were experimentally compared.


LIMITATIONS
-----------
The character-level model learns words one character at a time. It may
therefore produce realistic-looking words and Shakespeare-like formatting
without maintaining semantic consistency over a long passage.

Long-range dependencies remain challenging for recurrent models,
especially when the generated passage becomes much longer than the
training sequence.


POTENTIAL IMPROVEMENTS
----------------------
1. Attention mechanism
2. Three-layer LSTM
3. Larger hidden state
4. Larger training corpus
5. Word-level language model
6. Transformer comparison
7. Hyperparameter optimization
8. Beam-search generation
9. Web-based real-time generation interface
"""


with open(
    os.path.join(
        PROJECT_DIR,
        "analysis_report.txt"
    ),
    "w",
    encoding="utf-8"
) as file:

    file.write(
        analysis_text
    )


# ======================================================================
# STEP 66: CREATE FILE MANIFEST
# ======================================================================

manifest = []


for root, directories, files in os.walk(
    PROJECT_DIR
):

    for filename in files:


        filepath = os.path.join(
            root,
            filename
        )


        relative_path = os.path.relpath(
            filepath,
            PROJECT_DIR
        )


        size_mb = (

            os.path.getsize(
                filepath
            )

            / (
                1024
                * 1024
            )
        )


        manifest.append(
            {

                "File":
                    relative_path,

                "Size_MB":
                    round(
                        size_mb,
                        3
                    )
            }
        )


manifest_df = pd.DataFrame(
    manifest
)


manifest_df.to_csv(
    os.path.join(
        PROJECT_DIR,
        "file_manifest.csv"
    ),
    index=False,
    encoding="utf-8"
)


# ======================================================================
# STEP 67: PROJECT FOLDER SIZE
# ======================================================================

def get_folder_size_mb(
    folder
):

    total_size = 0


    for root, dirs, files in os.walk(
        folder
    ):

        for filename in files:


            filepath = os.path.join(
                root,
                filename
            )


            if os.path.isfile(
                filepath
            ):

                total_size += os.path.getsize(
                    filepath
                )


    return (

        total_size

        / (
            1024
            * 1024
        )
    )


folder_size = get_folder_size_mb(
    PROJECT_DIR
)


print(
    "\n"
    + "=" * 75
)

print(
    "PROJECT SIZE CHECK"
)

print(
    "=" * 75
)


print(
    f"Current Project Folder Size: "
    f"{folder_size:.2f} MB"
)


# ======================================================================
# STEP 68: CREATE SUBMISSION ZIP
# ======================================================================

ZIP_PATH = os.path.join(

    os.path.dirname(
        PROJECT_DIR
    ),

    "AI_Text_gen_submission.zip"
)


# Delete previous ZIP
if os.path.exists(
    ZIP_PATH
):

    os.remove(
        ZIP_PATH
    )


print(
    "\nCreating submission ZIP..."
)


with zipfile.ZipFile(

    ZIP_PATH,

    "w",

    compression=zipfile.ZIP_DEFLATED,

    compresslevel=9

) as zip_file:


    for root, directories, files in os.walk(
        PROJECT_DIR
    ):


        for filename in files:


            filepath = os.path.join(
                root,
                filename
            )


            archive_name = os.path.relpath(

                filepath,

                os.path.dirname(
                    PROJECT_DIR
                )
            )


            zip_file.write(
                filepath,
                archive_name
            )


# ======================================================================
# STEP 69: CHECK ZIP SIZE
# ======================================================================

zip_size = (

    os.path.getsize(
        ZIP_PATH
    )

    / (
        1024
        * 1024
    )
)


print(
    "\nZIP created:"
)

print(
    ZIP_PATH
)


print(
    f"\nZIP Size: "
    f"{zip_size:.2f} MB"
)


if zip_size < 50:

    print(
        "\nSUCCESS:"
    )

    print(
        "Submission ZIP is below 50 MB."
    )

else:

    print(
        "\nWARNING:"
    )

    print(
        "ZIP is larger than 50 MB."
    )

    print(
        "Delete unnecessary intermediate checkpoints "
        "before submitting."
    )


# ======================================================================
# STEP 70: FINAL RESULTS
# ======================================================================

print(
    "\n"
    + "=" * 75
)

print(
    "FINAL RESULTS"
)

print(
    "=" * 75
)


print(
    "\nLSTM RESULTS"
)

print(
    f"Validation Loss     : "
    f"{lstm_val_loss:.4f}"
)

print(
    f"Validation Accuracy : "
    f"{lstm_val_accuracy:.4f}"
)

print(
    f"Perplexity          : "
    f"{lstm_perplexity:.4f}"
)


print(
    "\nGRU RESULTS"
)

print(
    f"Validation Loss     : "
    f"{gru_val_loss:.4f}"
)

print(
    f"Validation Accuracy : "
    f"{gru_val_accuracy:.4f}"
)

print(
    f"Perplexity          : "
    f"{gru_perplexity:.4f}"
)


print(
    "\nProject Folder:"
)

print(
    PROJECT_DIR
)


print(
    "\nSubmission ZIP:"
)

print(
    ZIP_PATH
)


print(
    f"\nSubmission ZIP Size: "
    f"{zip_size:.2f} MB"
)


print(
    "\n"
    + "=" * 75
)

print(
    "PROJECT COMPLETED SUCCESSFULLY"
)

print(
    "=" * 75
)
```
